```python
import pandas as pd 
```

```python
df_csv = pd.read_csv('../data/raw/data_2023-01-01_2026-04-20.csv')

# Caculate cutting point
n = len(df_csv)
train_end = int(n * 0.8)
val_end = int(n * 0.9)

# Cutting data (Temporal split)
df_train = df_csv.iloc[:train_end]
df_val = df_csv.iloc[train_end:val_end]
df_test = df_csv.iloc[val_end:]

# Save as 3 diff files
df_train.to_parquet('../data/raw/data_train.parquet', engine='pyarrow', compression='snappy')
df_val.to_parquet('../data/raw/data_val.parquet', engine='pyarrow', compression='snappy')
df_test.to_parquet('../data/raw/data_test.parquet', engine='pyarrow', compression='snappy')

print(f"--- Data Splitting Completed ---")
print(f"Train samples: {len(df_train)}")
print(f"Val samples: {len(df_val)}")
print(f"Test samples: {len(df_test)}")
```
